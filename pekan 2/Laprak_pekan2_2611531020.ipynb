## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub mengulas fondasi penggunaan bahasa pemrograman Python, Visual Studio Code sebagai editor kode, serta GitHub sebagai platform untuk menyimpan dan mengelola kode program.

Praktikum Tipe Data dan Variabel mengupas konsep fundamental dalam bahasa pemrograman Python yang berfungsi untuk menyimpan dan memproses data. Tipe data menentukan kategori nilai yang dipakai dalam program, sementara variabel berperan sebagai wadah untuk menyimpan atau mereferensikan nilai-nilai tersebut.

1. Integer (int)
Integer adalah tipe data yang berfungsi untuk menampung bilangan bulat, contohnya 10, 25, dan -5.
2. Float (float)
Float merupakan tipe data yang dipakai untuk menyimpan bilangan yang mengandung nilai desimal, seperti 3.14, 10.5, dan 0.75.
3. Complex (complex)
Complex adalah tipe data yang digunakan untuk menampung bilangan kompleks yang terdiri dari bagian real dan imajiner. Sebagai contoh adalah 3+2j.
4. String (str)
String merupakan tipe data yang berfungsi untuk menyimpan teks atau rangkaian karakter. String ditulis dengan menggunakan tanda kutip, misalnya "Python" dan "Informatika".
5. Boolean (bool)
Boolean adalah tipe data yang hanya memiliki dua kemungkinan nilai, yaitu True dan False. Tipe data ini umumnya dipakai untuk merepresentasikan kondisi benar atau salah.
6. Konstanta
Konstanta adalah nilai yang secara konsep tidak dirancang untuk diubah sepanjang program berjalan. Python tidak menyediakan sintaks khusus untuk mendefinisikan konstanta, namun penamaan dengan huruf kapital dapat digunakan sebagai indikator bahwa suatu variabel diperlakukan sebagai konstanta, contohnya PI = 3.14.
7. Variabel
Variabel berfungsi sebagai wadah untuk menyimpan atau mereferensikan sebuah nilai. Dalam Python, variabel dapat didefinisikan dengan memberikan nilai menggunakan operator =, misalnya nama = "Varren Della" atau umur = 18.
8. Input dan Output
Python menyediakan fungsi input() untuk menerima masukan dari pengguna dan print() untuk menampilkan data ke layar. Data yang diterima dari input() secara default bertipe str, sehingga perlu dilakukan konversi jika data tersebut akan digunakan sebagai bilangan.

### 2.2.5 Program Konstanta_2611531020
</div>

```python
# Buat file dengan nama Konstanta_2611531021
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_1234 = float(input("Masukkan nilai jari-jari: "))
luas_1234 = PI * jari_1234
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1234,luas_1234))
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

3. Cetak ("Pi: %f" % (PI))

5. Nilai luas_1234_ = PI * jari_1234_
